#### Q12.3: How many TV shows got the 'R' rating between the minimum and maximum year? Make sure the data-type and format are correct.

```python
# finding the minimum year in the dataset

dataset['Year'].min()
```

```python
# finding the maximum year in the dataset

dataset['Year'].max()
```

```python
# removing the null values to avoid error during conversion of data-type
# changing the data-type to integer
# viewing the new_dataset

new_dataset = dataset.dropna()
new_dataset['Year'] = new_dataset['Year'].astype(int)
new_dataset
```

```python
# using comparison operators for "between" values

new_dataset[(new_dataset['Category'] == 'TV Show') & (new_dataset['Rating'] == 'R') & (2021 >= new_dataset['Year']) & (new_dataset['Year'] >= 2008)].shape
```

```python
# using the between() method

new_dataset[(new_dataset['Category'] == 'TV Show') & (new_dataset['Rating'] == 'R') & (new_dataset['Year'].between(2008, 2021))].shape
```
